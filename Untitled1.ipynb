Задание А второго тура

```python
import csv
import numpy as np

item_ids = []
categories = []
prices = []
ratings = []
brands = []
in_stocks = []

with open('items_A.csv', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        item_ids.append(int(row['item_id']))
        categories.append(row['category'])
        prices.append(int(row['price']))
        ratings.append(float(row['rating']))
        brands.append(row['brand'])
        in_stocks.append(int(row['in_stock']))

N = len(item_ids)

categories_np = np.array(categories)
prices_np = np.array(prices)
ratings_np = np.array(ratings)
in_stocks_np = np.array(in_stocks)
```

```python
# A1
mask_A1 = (categories_np == 'phones') & (ratings_np >= 4.5) & (in_stocks_np == 1)
print(np.sum(mask_A1))
```

```python
# A2
sum_price = {}
cnt_brand = {}
for i in range(N):
    if categories[i] == 'laptops':
        b = brands[i]
        if b not in sum_price:
            sum_price[b] = 0
            cnt_brand[b] = 0
        sum_price[b] += prices[i]
        cnt_brand[b] += 1

best_brand = None
best_avg = -1.0
for b in sum_price:
    avg = sum_price[b] / cnt_brand[b]
    if avg > best_avg:
        best_avg = avg
        best_brand = b
print(best_brand)
```

```python
# A3
mask_A3 = (in_stocks_np == 1) & (ratings_np >= 4.5) & (prices_np >= 50000)
print(np.sum(mask_A3))
```

Жестко готовимся к А4 и А5 и считываем граф

```python
item_ids = []
categories = []

with open('items_A.csv', encoding='utf-8') as f:
    for row in csv.DictReader(f):
        item_ids.append(int(row['item_id']))
        categories.append(row['category'])

edges_from = []
edges_to = []
with open('also_viewed_A.csv', encoding='utf-8') as f:
    for row in csv.DictReader(f):
        edges_from.append(int(row['item_from']))
        edges_to.append(int(row['item_to']))

N = len(item_ids)
M = len(edges_from)
```

```python
# A4
item_to_set = set()
for i in range(M):
    item_to_set.add(edges_to[i])

cnt = {}
for i in range(N):
    cat = categories[i]
    if cat not in cnt:
        cnt[cat] = 0
    if item_ids[i] in item_to_set:
        cnt[cat] += 1

cats = list(cnt.keys())
cats.sort()

with open('answer4.csv', 'w', encoding='utf-8', newline='') as f:
    writer = csv.writer(f)
    writer.writerow(['category', 'cnt'])
    for c in cats:
        writer.writerow([c, cnt[c]])

print(cnt)
```

```python
# A5
N_ID = 501

g = []
for i in range(N_ID):
    g.append([])
used = [False] * N_ID

for i in range(M):
    a = edges_from[i]
    b = edges_to[i]
    g[a].append(b)
    g[b].append(a)

cat_by_id = [''] * N_ID
for i in range(N):
    cat_by_id[item_ids[i]] = categories[i]

have_p = False
have_a = False

def dfs(v):
    global have_p, have_a
    used[v] = True
    if cat_by_id[v] == 'phones':
        have_p = True
    if cat_by_id[v] == 'accessories':
        have_a = True
    for u in g[v]:
        if not used[u]:
            dfs(u)

ans = 0
for i in range(N):
    v = item_ids[i]
    if not used[v]:
        have_p = False
        have_a = False
        dfs(v)
        if have_p and have_a:
            ans += 1

print(ans)
```

Задача Д

```python
points = [
    (0,8), (0,7), (0,2), (0,0),
    (4,7), (4,4),
    (6,5), (6,1),
    (9,7), (9,4), (9,1), (9,0),
    (11,6), (11,2)
]

def median(arr):
    a = arr[:]
    a.sort()
    return a[len(a) // 2]

best_mae = 1e18

a = -2.0
while a <= 2.0:
    z = []
    for (x, y) in points:
        z.append(y - a * x)
    b = median(z)
    
    mae = 0.0
    for (x, y) in points:
        mae += abs(a * x + b - y)

    if mae < best_mae:
        best_mae = mae

    a += 0.001
    
print(best_mae/14.0)
```
